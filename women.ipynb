1.IMPORT LIBRARIES

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestRegressor
```

2.LOAD DATASET

```python
df = pd.read_csv("KIET_Women_Safety_365_Days_2023_2026 (1).csv")
```

4.CHECKING NULL VALUES

```python
df.isnull().sum()
```

```python
df["Lighting_Score"] = df["Lighting_Score"].fillna(
    df["Lighting_Score"].median()
)
```

5.CHECK AND REMOVE DUPLICATE ROWS, COLUMNS

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

6.CONVERTING DATE COLUMN

```python
df["Date"] = pd.to_datetime(df["Date"])
```

7.CHECK CATEGORICAL COLUMNS

```python
df["Place"].unique()
```

```python
df["Time_Period"].unique()
```

8.ENCODE CATEGORICAL DATA

```python
df = pd.get_dummies(
    df,
    columns=["Place", "Time_Period"],
    drop_first=True
)
```

9.SELECT INPUT FEATURES

```python
X = df[
    [
        "Lighting_Score",
        "CCTV_Available",
        "Crowd_Density",
        "Security_Nearby",
        "Synthetic_Incident_Index"
    ]
]
```

10.SELECT TARGET

```python
y = df["Risk_Score"]
```

11. TRAIN-SPLIT THE DATA

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

12. FEATURE SCALING

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled[:5]
```

13.CREATE POLYNOMIAL FEATURE

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2)
X_train_poly = poly.fit_transform(X_train_scaled)
X_test_poly = poly.transform(X_test_scaled)
```
